### Data pre-processing

All column names were converted to lowercase using this to ensure consistency and simplify data handling during the analysis process.

```python

df.columns=df.columns.str.lower()
```

The column was converted into datetime format to enable time-based analysis. Missing values were handled using the forward fill `fillna()`method to maintain data continuity.

```python
df['order_date']=pd.to_datetime(df['order_date'],format='%d-%m-%Y')
df['order_date']=df['order_date'].fillna(method='ffill')
```

Rows with missing values in the **size** column were removed to ensure that the dataset contains only valid product size information for analysis.

```python
df = df.dropna(subset=['size'])
```

The **mrp** column was renamed to **price** to make the dataset more clear and consistent for analysis.

```python
df=df.rename(columns={'mrp':'price'})
```

Rows with missing values in both price and **units_sold** columns were removed to maintain the accuracy of the sales analysis.

```python
 df= df[~((df['price'].isna()) & (df['units_sold'].isna()))]
```

Missing values in the **price** column were replaced with the average price of the respective product using group-based mean imputation.

```python
df['price'] = df.groupby('product_name')['price'].transform(lambda x: x.fillna(x.mean()))
```

Rows with missing values in the **units_sold** column were removed to maintain the accuracy of the sales data.

```python
df=df.dropna(subset=["units_sold"])
```

Missing values in the **discount_applied** column were replaced with 0 using the pandas `fillna()` function. This assumes that if the discount value is missing, no discount was applied to the product.

```python
df['discount_applied']=df['discount_applied'].fillna('0')
```

The columns **unnamed: 13** and **unnamed: 14** were removed from the dataset using the pandas `drop()` function because they did not contain meaningful information and were likely created during data import or file formatting.

```python
df = df.drop(columns=['unnamed: 13','unnamed: 14'])
```

The dataset was filtered to identify records where the **size** column contains numeric values using the `str.isnumeric()` function. The filtered data was then grouped by product_name to examine the different numeric sizes available for each product. This helps understand size variations across products, especially for items like footwear where sizes are typically numeric.

```python
numeric_sizes = df[df['size'].str.isnumeric()]
print(numeric_sizes.groupby('product_name')['size'].unique())
```

The dataset was filtered to identify records where the **size** column contains non-numeric (text) values by converting the column to string format and using `str.isnumeric()` with a negation condition. The filtered data was then grouped by product_name to observe the different text-based sizes available for each product, such as apparel sizes like S, M, L, and XL

```python
text_sizes = df[~df['size'].astype(str).str.isnumeric()]
print(text_sizes.groupby('product_name')['size'].unique())
```

A mapping dictionary was created to convert text-based apparel sizes (S, M, L, XL) into numeric values. A function named `convert_to_numeric()` was then defined to standardize the size column by converting both text-based and numeric sizes into a consistent numeric format. This allows easier analysis and comparison of product sizes across the dataset.

```python
# Map for letter sizes
size_map = {"S":1, "M":2, "L":3, "XL":4}

# Function to convert to numeric
def convert_to_numeric(size_list):
    numeric_list = []
    for s in size_list:
        if str(s) in size_map:   # letter size
            numeric_list.append(size_map[str(s)])
        else:                    # numeric size
            numeric_list.append(int(s))
    return numeric_list

```

After converting the size values into numeric format, the datatype of the **size** column was changed to integer using the `astype()` function. This ensures that the column can be used effectively for numerical analysis.

```python
df['size'] = convert_to_numeric(df['size'])
df['size'] = df['size'].astype(int)
```

The **region** column contained inconsistent city names due to variations in spelling and letter casing (e.g.,"hyd", "hyderabad", "hyderbad", "bengaluru", etc.). To ensure consistency, all region names were first converted to lowercase using the str.lower() function. A mapping dictionary was then created to standardize different variations of city names into a single consistent format using the replace() function.

```python
df["region"] = df["region"].str.lower()

region_map = {
    "hyd": "Hyderabad",
    "hyderabad": "Hyderabad",
    "hyderbad": "Hyderabad",
    "bangalore": "Bangalore",
    "bengaluru": "Bangalore",
    "mumbai": "Mumbai",
    "delhi": "Delhi",
    "pune": "Pune",
    "kolkata": "Kolkata"
}

df["region"] = df["region"].replace(region_map)
```
